```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.spatial.transform import Rotation

from system.optical_system import OpticalSystem

from optics.mirror import Mirror
from optics.beamsplitter import BeamSplitter
from optics.window import Window
from optics.plate_beamsplitter import PlateBeamSplitter
from optics.detector import Detector

from optics.bundle_source import CircularBundleSource

from optics.zernike_figure_error import (
    ZernikeFigureError
)

from plotting.views import plot_system_3views

from materials.constant_index import (
    ConstantIndex
)

bk7 = ConstantIndex(
    1.5168
)

# =====================================================
# PARAMETERS
# =====================================================

wavelength = 632.8e-9

bundle_radius = 0.1
bundle_n_rays = 1


# =====================================================
# WAVEFRONT ERRORS
# =====================================================

mirror1_error = ZernikeFigureError.mirror_lambda_pv(
        wavelength=wavelength,
        ratio=4,
        pupil_radius=15,
    )

mirror2_error = ZernikeFigureError.mirror_lambda_pv(
        wavelength=wavelength,
        ratio=4,
        pupil_radius=15,
    )

bs_error = (
    ZernikeFigureError
    .beamsplitter_lambda_pv(
        wavelength=wavelength,
        ratio=2,
        pupil_radius=15,
    )
)

# =====================================================
# SYSTEM
# =====================================================

system = OpticalSystem()


# =====================================================
# MIRRORS
# =====================================================

mirror1 = Mirror(
    name="M1",

    position=[100, 0, 100],

    rotation=Rotation.identity(),

    width=30,
    height=30,

    wavefront_error=mirror1_error,
)

mirror2 = Mirror(
    name="M2",

    position=[250, 0, 0],

    rotation=Rotation.from_euler(
        "y",
        -90,
        degrees=True,
    ),

    width=30,
    height=30,

    wavefront_error=mirror2_error,
)


# =====================================================
# BEAM SPLITTER
# =====================================================

bs = PlateBeamSplitter(

    name="BS",

    position=[100,0,0],

    rotation=Rotation.from_euler(
        "y",
        +135,
        degrees=True,
    ),

    thickness=10,

    material=bk7,

    width=50,
    height=50,

    R=0.5,
    T=0.5,
)

### GLASS ###################

glass = Window(
    name="GLASS",

    thickness=10.0,

    material=ConstantIndex(
        1.5
    ),

    position=[100,0,0],

    rotation=Rotation.from_euler(
        "y",
        -45,
        degrees=True,
    ),

    width=50,
    height=50,
)

# =====================================================
# DETECTORS
# =====================================================

detector1 = Detector(
    name="DET1",

    position=[-10, 0, 0],

    rotation=Rotation.from_euler(
        "y",
        -90,
        degrees=True,
    ),

    width=30,
    height=30,
)

detector2 = Detector(
    name="DET2",

    position=[100, 0, -100],

    rotation=Rotation.identity(),

    width=30,
    height=30,
)


# =====================================================
# SOURCE
# =====================================================

source = CircularBundleSource(
    position=[0,0,0],

    direction=[1,0,0.0],

    wavelength=wavelength,

    radius=bundle_radius,

    n_rays=bundle_n_rays,
)


# =====================================================
# ASSEMBLE SYSTEM
# =====================================================

#bs.add_to_system(
#    system
#)
#system.add_element(glass)

system.add_element(mirror1)
system.add_element(mirror2)

system.add_detector(detector1)
system.add_detector(detector2)

system.set_source(source)


# =====================================================
# ONE SINGLE TRACE
# =====================================================

system.randomize_wavefront_errors()

last_rays = system.trace()


# =====================================================
# INTERFEROGRAM
# =====================================================

opd_values = np.linspace(
    -5*wavelength,
     5*wavelength,
    1000,
)

signal_det1 = system.interferogram(
    opd_values,
    detector1,
)

signal_det2 = system.interferogram(
    opd_values,
    detector2,
)


# =====================================================
# MODULATION EFFICIENCY
# =====================================================

Imax = np.max(
    signal_det2
)

Imin = np.min(
    signal_det2
)

ME = (
    Imax - Imin
) / (
    Imax + Imin
)


# =====================================================
# RESULTS
# =====================================================

print()
print("===================================")
print(" INTERFEROGRAM")
print("===================================")
print()

print(
    f"Bundle rays : {bundle_n_rays}"
)

print(
    f"ME : {ME:.6f}"
)

print()

detector1.report()

print()

detector2.report()


# =====================================================
# INTERFEROGRAM PLOT
# =====================================================

plt.figure(
    figsize=(10,5)
)

plt.plot(
    opd_values / wavelength,
    signal_det1,
    label="DET1",
)

plt.plot(
    opd_values / wavelength,
    signal_det2,
    label="DET2",
)

plt.xlabel(
    "Artificial OPD / λ"
)

plt.ylabel(
    "Intensity"
)

plt.grid()

plt.legend()

plt.tight_layout()

plt.show()


# =====================================================
# GEOMETRY VIEW
# =====================================================

plot_system_3views(
    system.tracer,
    last_rays,
)


mirror1_error.plot()

print(mirror1_error.statistics())

mirror2_error.plot()

print(mirror2_error.statistics())


bs_error.plot()

print(bs_error.statistics())
```

```python
import numpy as np

from scipy.spatial.transform import Rotation

from system.optical_system import OpticalSystem

from optics.window import Window
from optics.detector import Detector
from optics.source import Source

from materials.constant_index import ConstantIndex

from plotting.views import plot_system_3views


# =====================================================
# SYSTEM
# =====================================================

system = OpticalSystem()


# =====================================================
# MATERIAL
# =====================================================

glass = ConstantIndex(
    1.5
)


# =====================================================
# WINDOW
# =====================================================

window = Window(
    name="WINDOW",

    thickness=10.0,

    material=glass,

    position=[100,0,0],

    rotation=Rotation.from_euler(
        "y",
        -45,
        degrees=True,
    ),

    width=50,
    height=50,
)


# =====================================================
# DETECTOR
# =====================================================

detector = Detector(
    name="DET",

    position=[300,0,0],

    rotation=Rotation.from_euler(
        "y",
        -90,
        degrees=True,
    ),

    width=100,
    height=100,
)


# =====================================================
# SOURCE
# =====================================================

source = Source(
    position=[0,0,0],

    direction=[1,0,0],

    wavelength=632.8e-9,
)


# =====================================================
# BUILD SYSTEM
# =====================================================

system.add_element(
    window
)

system.add_detector(
    detector
)

system.set_source(
    source
)


# =====================================================
# TRACE
# =====================================================

rays = system.trace()

print()

for ray in rays:

    ray.summary()

print()

detector.report()


# =====================================================
# DETECTOR SPOT
# =====================================================

for hit in detector.hits:

    print(
        "Detector position:",
        hit["position"]
    )


# =====================================================
# GEOMETRY
# =====================================================

plot_system_3views(
    system.tracer,
    rays,
)
```

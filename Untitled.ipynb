```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.spatial.transform import Rotation

from system.optical_system import OpticalSystem

from optics.corner_cube import CornerCube
from optics.compensator import Compensator
from optics.plate_beamsplitter import PlateBeamSplitter
from optics.detector import Detector
from optics.bundle_source import CircularBundleSource

from plotting.views import plot_system_3views

from materials.constant_index import ConstantIndex


# =====================================================
# PARAMETERS
# =====================================================

wavelength = 632.8e-9

bk7 = ConstantIndex(
    1.5168,
    "BK7",
)

arm_length = 150

# =====================================================
# SYSTEM
# =====================================================

system = OpticalSystem()

# =====================================================
# SOURCE
# =====================================================

source = CircularBundleSource(

    position=[-50,-10,10],

    direction=[1,0,0],

    wavelength=wavelength,

    radius=0.1,

    n_rays=1,
)

system.set_source(
    source
)

# =====================================================
# BEAMSPLITTER
# =====================================================

bs = PlateBeamSplitter(

    name="BS",

    position=[0,0,0],

    rotation=Rotation.from_euler(
        "y",
        135,
        degrees=True,
    ),

    thickness=10,

    width=50,
    height=50,

    material=bk7,

    R=0.5,
    T=0.5,
)

bs.add_to_system(
    system
)

# =====================================================
# COMPENSATOR
# =====================================================

comp = Compensator(

    name="COMP",

    position=[0,0,50],

    rotation=Rotation.from_euler(
        "y",
        135,
        degrees=True,
    ),

    thickness=10,

    width=50,
    height=50,

    material=bk7,
)

comp.add_to_system(
    system
)

# =====================================================
# CORNER CUBE 1
# =====================================================

cc1 = CornerCube(

    name="CC1",

    position=[0,0,arm_length],

    size=40,

    optical_axis=[0,0,1],
)

cc1.add_to_system(
    system
)

# =====================================================
# CORNER CUBE 2 (mobile)
# =====================================================

cc2 = CornerCube(

    name="CC2",

    position=[arm_length,0,0],

    size=40,

    optical_axis=[1,0,0],
)

cc2.add_to_system(
    system
)

# =====================================================
# DETECTORS
# =====================================================

det1 = Detector(

    name="DET1",

    position=[-100,0,0],

    rotation=Rotation.from_euler(
        "y",
        -90,
        degrees=True,
    ),

    width=50,
    height=50,
)

system.add_detector(
    det1
)

det2 = Detector(

    name="DET2",

    position=[0,0,-100],

    rotation=Rotation.identity(),

    width=50,
    height=50,
)

system.add_detector(
    det2
)

# =====================================================
# FIRST TRACE
# =====================================================

rays = system.trace()

print()
print("====================================")
print("CORNER CUBE MICHELSON")
print("====================================")

print()

det1.report()

print()

det2.report()

# =====================================================
# GEOMETRY
# =====================================================

plot_system_3views(
    system.tracer,
    rays,
)

# =====================================================
# PHYSICAL INTERFEROGRAM
# =====================================================

scan = np.linspace(
    -3*wavelength,
     3*wavelength,
    201,
)

signal_det1 = []
signal_det2 = []

zpd_position = (
    cc2.position.copy()
)

for dx in scan:

    cc2.position = (
        zpd_position
        +
        np.array([
            dx,
            0,
            0,
        ])
    )

    system.trace()

    signal_det1.append(
        det1.intensity
    )

    signal_det2.append(
        det2.intensity
    )

# restore

cc2.position = (
    zpd_position
)

signal_det1 = np.asarray(
    signal_det1
)

signal_det2 = np.asarray(
    signal_det2
)

opd = 2*scan

# =====================================================
# VISIBILITY
# =====================================================

def visibility(signal):

    Imax = np.max(signal)
    Imin = np.min(signal)

    return (
        (Imax - Imin)
        /
        (Imax + Imin)
    )

print()

print(
    "Visibility DET1 =",
    visibility(signal_det1)
)

print(
    "Visibility DET2 =",
    visibility(signal_det2)
)

# =====================================================
# PLOTS
# =====================================================

plt.figure(
    figsize=(10,5)
)

plt.plot(
    opd/wavelength,
    signal_det1,
    label="DET1",
)

plt.plot(
    opd/wavelength,
    signal_det2,
    label="DET2",
)

plt.xlabel(
    "OPD / λ"
)

plt.ylabel(
    "Intensity"
)

plt.title(
    "Corner Cube Michelson"
)

plt.grid()

plt.legend()

plt.tight_layout()

plt.show()

# =====================================================
# ENERGY CONSERVATION
# =====================================================

plt.figure(
    figsize=(8,4)
)

plt.plot(
    opd/wavelength,
    signal_det1 + signal_det2,
)

plt.title(
    "DET1 + DET2"
)

plt.xlabel(
    "OPD / λ"
)

plt.ylabel(
    "Intensity"
)

plt.grid()

plt.tight_layout()

plt.show()
```
